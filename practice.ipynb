## if statement

```python
if<condition>:
    <code>
elif<condition>:
    <code>

else<condition>:
    <code>

```

```python
age=16

if age>=18:
    print('voter')
elif age>=15:
    print('teenager')
else:
    print('no voter')
```

## for loop

```python
for<variable> in <list>:
    <code>
```

```python
coutries=['uganda','kenya','rwanda','tanzania']
```

```python
for country in coutries:
    print(country)
```
