```python
#while and if statement
age_str = input("enter_the_age:")
age = int(age_str) # Convert input to integer

while age <= 15:
  if age > 14:
    print("senior student")
  elif age >= 12:
    print("junior student")
  else: # age < 12
    print("too young")
  age += 1 # Increment age to avoid infinite loop

else: # This else belongs to the while loop and executes when the while condition is false (age > 15)
  print("not in range")
```
